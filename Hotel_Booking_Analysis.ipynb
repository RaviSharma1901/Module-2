````python
# Dataset Duplicate Value Count
def duplicate_val_cnt(df):
  try:
      return print('Count of Duplicate Value is {}'.format(df.duplicated().sum()))
  except FileNotFoundError:
      print("Error: The file was not found. Please check the file path and try again.")
  except Exception as e:
      print("An error occurred:", {e})
duplicate_val_cnt(hotel_df)
````

#### Missing Values/Null Values

````python
# function to find Missing Values/Null Values Count
def missing_val_cnt(df):
  try:
      null_val = df.isnull().sum()
      # reset the index and rename the column
      after_reset_col = null_val.reset_index().rename(columns={'index':'Column',0:'Null Values'})
      return after_reset_col
  except Exception as e:
      print("An error occurred:", {e})

# get Missing Values/Null Values count
missing_val_cnt(hotel_df)
````

````python
# Visualizing the missing values
def missing_val_visual(df):
  try:
      plt.figure(figsize=(10,5))
      sns.heatmap(df.isnull(),cbar=False)
      plt.title('Missing Values Heatmap')
      plt.xlabel("Column Labels")
      plt.ylabel("Row Index")
      plt.show()
  except Exception as e:
      print("An error occurred:", {e})
missing_val_visual(hotel_df)
````

### Data Wrangling Code

````python
# Write your code to make your dataset analysis ready.

# make a copy of dataframe to work on df
df_copy = hotel_df.copy()
````

````python
# Deleting Duplicates values from dataset
df_copy.drop_duplicates(inplace=True)

# Getting count of rows and columns after removed duplicates
print(f'the number of rows: {df_copy.shape[0]} and columns: {df_copy.shape[1]} after removed duplicates')
````

````python
# Calculate the percentage of null values of each column

# function to get percentage of null values
def cal_null_val_percentage(df):
    try:
        # Calculate the percentage of null values
        percent = df.isna().sum()/df.shape[0]*100     # df.isna() or df.isnull()

        # sorting null values in descending order and reset the index and rename the column
        perc_srt = percent.sort_values(ascending=False).reset_index().rename(columns={'index':'Columns',0:'Null %'})

        return perc_srt
    except Exception as e:
        print("An error occurred:", {e})
cal_null_val_percentage(df_copy)
````

````python
# In above result we can see that column company has approx. 94% of missing value
# which is very high so droping the column company
df_copy.drop(columns=['company'],inplace=True)
````

````python
# Now check for the Other columns they have less % of null values so replace them with 0

df_copy.fillna({'agent':0,'country':'Others','children':0},inplace=True)
````

#### Chart - 3
#### **Analyze the monthly arrival patterns of guests to identify the busiest, moderately busy, and least occupied months.**

````python
# Chart - 3 visualization code

# get the Arrival Month and No
arrival_monthly_guest_count = df_copy['arrival_date_month'].value_counts().reset_index()
arrival_monthly_guest_count.columns = ['Arrival Month', 'No of guests']

# month order
month_order = ['January', 'February', 'March', 'April', 'May', 'June', 'July', 'August', 'September', 'October', 'November', 'December']

# Convert 'Arrival Month' to Categorical type with custom month order
arrival_monthly_guest_count['Arrival Month'] = pd.Categorical(arrival_monthly_guest_count['Arrival Month'], categories=month_order, ordered=True)

# Sort by the defined month order
arrival_monthly_guest_count = arrival_monthly_guest_count.sort_values('Arrival Month')

# Find the max guest month
max_guest_index = arrival_monthly_guest_count['No of guests'].idxmax() # 0
max_guest_month = arrival_monthly_guest_count.loc[max_guest_index, 'Arrival Month'] # August


plt.figure(figsize=(10, 6))
plt.plot(arrival_monthly_guest_count['Arrival Month'], arrival_monthly_guest_count['No of guests'], marker='o', linestyle='-', color='b')

# Highlight the busiest month in red
plt.plot(arrival_monthly_guest_count['Arrival Month'][max_guest_index], arrival_monthly_guest_count['No of guests'][max_guest_index], 'ro')  # 'ro' - red color marker

# Labels and title
plt.xlabel('Arrival Month')
plt.ylabel('Number of Guests')
plt.title('Monthly Arrival Pattern of Guests')
plt.tight_layout()
plt.show()
````

#### Chart - 14 - **Correlation Heatmap**

````python
# Correlation Heatmap visualization code
corr_df = df_copy[['lead_time', 'previous_cancellations', 'previous_bookings_not_canceled', 'booking_changes', 'days_in_waiting_list','adr',
          'required_car_parking_spaces','total_of_special_requests']]

# calculates the correlation matrix. (1 = perfect positive correlation, -1 = perfect negative correlation, 0 = no correlation)
corr_matrix=corr_df.corr()

plt.figure(figsize=(10,5))

sns.heatmap(corr_matrix, vmin=-1,annot=True,cmap='coolwarm') # vmin=1 - minimum value of the heatmap color scale to -1
````

##### 1. Why did you pick the specific chart?

**By using a heatmap to visualize the correlation matrix, you gain a comprehensive overview of how different features relate to each other, enabling more informed decisions in your analysis and strategy development.**

**Color Scale:**

***Red Colors:*** Represent positive correlations. The more intense the red, the stronger the positive correlation.

***Blue Colors:*** Represent negative correlations. The more intense the blue, the stronger the negative correlation.

***White/Neutral Colors:*** Represent correlations close to zero, indicating no strong relationship.

**Annotations:** Each cell is annotated with the correlation value, making it easy to see the exact correlation between pairs of features.

**Diagonal:** The diagonal represents the correlation of each feature with itself, which is always 1 (perfect correlation).

**vmin=-1:** Sets the minimum value of the heatmap color scale to -1.

**annot=True:** Annotates each cell with the numeric value of the correlation.

**cmap='coolwarm':** where cool colors (blue) represent negative correlations and warm colors (red) represent positive correlations.

##### 2. What is/are the insight(s) found from the chart?

**Lead Time vs. Previous Cancellations:** If the correlation is negative, it might indicate that guests with longer lead times have fewer previous cancellations.

**Days in Waiting List vs. Required Car Parking Spaces:** A low correlation (close to zero) suggests that these two features are not related.

**ADR (Average Daily Rate) vs. Total of Special Requests:** A positive correlation suggests that higher ADRs might be associated with more special requests.
